# 03 EDA and Business Analysis

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../data/processed/walmart_cleaned.csv", parse_dates=["Date"])
```

```
store = df.groupby("Store")["Weekly_Sales"].agg(["sum","mean","std"]).sort_values("sum", ascending=False)
store.head(10)
```

```
monthly = df.groupby("Year_Month")["Weekly_Sales"].sum()
monthly.plot(figsize=(12,5), title="Monthly Walmart Sales")
plt.xticks(rotation=60)
plt.show()
```

```
corr = df[["Weekly_Sales","Temperature","Fuel_Price","CPI","Unemployment"]].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.show()
```
