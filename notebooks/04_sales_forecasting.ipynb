# 04 Sales Forecasting

```
import pandas as pd
from statsmodels.tsa.holtwinters import ExponentialSmoothing
df = pd.read_csv("../data/processed/walmart_cleaned.csv", parse_dates=["Date"])
weekly = df.groupby("Date")["Weekly_Sales"].sum().sort_index()
weekly.tail()
```

```
test_size = 12
train, test = weekly.iloc[:-test_size], weekly.iloc[-test_size:]
model = ExponentialSmoothing(
    train, trend="add", seasonal="add",
    seasonal_periods=52, initialization_method="estimated"
).fit()
forecast = model.forecast(test_size)
pd.DataFrame({"Actual":test.values, "Forecast":forecast.values}, index=test.index)
```
