# 02 Data Cleaning

```
import pandas as pd
from pathlib import Path
df = pd.read_csv(Path("../data/raw/Walmart_DataSet.csv"))
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
```

```
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Quarter"] = df["Date"].dt.quarter
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)
df["Holiday"] = df["Holiday_Flag"].map({0:"Non-Holiday", 1:"Holiday"})
df["Sales_Million"] = df["Weekly_Sales"] / 1_000_000
df.to_csv("../data/processed/walmart_cleaned.csv", index=False)
df.head()
```
